```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/kaggle/input/datasets/carrie1/ecommerce-data/data.csv",encoding="ISO-8859-1")
df.info()
df.describe()
df.head()
```

```python
#missing value check
print(df.isnull().sum())
```

```python
df["Description"]=df["Description"].fillna("UNKNOWN")
df["CustomerID"]=df["CustomerID"].fillna("UNKNOWN")


```

```python

print(df.isnull().sum())
```

```python
#Duplicate value check
print(df.duplicated().sum())
```

```python
#Remove duplicate value
df=df.drop_duplicates()
print("Shape after removing duplicates:",df.shape)
```

```python
# Convert columns to correct data types

df['InvoiceNo'] = df['InvoiceNo'].astype('string')

df['StockCode'] = df['StockCode'].astype('string')

df['Description'] = df['Description'].astype('string')

df['Quantity'] = pd.to_numeric(
    df['Quantity'], errors='coerce'
).astype('Int64')

df['InvoiceDate'] = pd.to_datetime(
    df['InvoiceDate'],
    errors='coerce',
    dayfirst=True
)

df['UnitPrice'] = pd.to_numeric(
    df['UnitPrice'], errors='coerce'
)

df['CustomerID'] = pd.to_numeric(
    df['CustomerID'], errors='coerce'
).astype('Int64')

df['Country'] = df['Country'].astype('string')

print("Data types converted successfully!")
```

```python

# INCONSISTENT VALUES CHECKING & CLEANING

# 1. Clean extra spaces in text columns
text_columns = ['Description', 'Country', 'StockCode']

for col in text_columns:
    df[col] = df[col].astype('string').str.strip()
    df[col] = df[col].str.replace(r'\s+', ' ', regex=True)

# 2. Standardize Country names
df['Country'] = df['Country'].str.title()

# 3. Check negative Quantity
negative_quantity = (df['Quantity'] < 0).sum()

# 4. Check zero and negative UnitPrice
zero_price = (df['UnitPrice'] == 0).sum()
negative_price = (df['UnitPrice'] < 0).sum()

# 5. Display unusual values before deciding what to do
print("========== INCONSISTENT VALUES REPORT ==========")

print("\nNegative Quantity:", negative_quantity)
print("Zero UnitPrice:", zero_price)
print("Negative UnitPrice:", negative_price)

print("\nCountry Names:")
print(df['Country'].unique())

# 6. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 7. Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# 8. Final dataset information
print("\nFinal Dataset Shape:", df.shape)

print("\nFinal Data Types:")
print(df.dtypes)

print("\nCleaning and checking completed!")
```

```python
import matplotlib.pyplot as plt
```

```python
#Top 10 Best Selling product
top_products=(df.groupby("Description")["Quantity"].sum().sort_values(ascending=False).head(10))
plt.figure(figsize=(12,6))
top_products.plot(kind="bar")
plt.title("Top 10 Best Selling Products")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=75)
plt.show()
'''Insight: Top 10 Best-Selling Products

WORLD WAR 2 GLIDERS ASSTD DESIGNS is the best-selling product, followed by JUMBO BAG RED RETROSPOT. These products
show strong customer demand.

Recommendation: Maintain sufficient stock of popular products to meet customer demand and avoid
stock shortages.'''
```

```python
#Top 10 Countries By Sales
df["TotalSales"]=df["Quantity"]+df["UnitPrice"]

country_sales = (df.groupby('Country')['TotalSales'].sum().sort_values(ascending=False).head(10))
country_sales.plot(kind='bar', figsize=(12,6))
plt.title('Top 10 Countries by Sales')
plt.ylabel('Total Sales')
plt.show()
'''Insight: Top 10 Countries by Sales

The United Kingdom generates the highest sales by a significant margin, while the Netherlands and
EIRE are the next major contributors. This indicates that the business is heavily dependent on the
UK market.

Recommendation: Focus on maintaining strong customer relationships in the UK while
exploring opportunities to increase sales in other countries.'''
```

```python
#Monthly Revenue Trend
df["InvoiceDATE"]=pd.to_datetime(df["InvoiceDate"],errors="coerce",dayfirst=True)
df["Month"]=df["InvoiceDATE"].dt.to_period("M")
monthly_sales=df.groupby("Month")["TotalSales"].sum()
monthly_sales.plot(figsize=(12,6),marker="o")
plt.title("Monthly Revenue Trend")
plt.xticks(rotation=45)
plt.show()
'''Insight: Monthly Revenue Trend

The monthly revenue shows an overall increasing trend from 2010 to 2011. Revenue rises significantly
in early 2011 and reaches its highest point around May 2011. However, some fluctuations occur in the following months.

Recommendation: Analyze peak sales months to understand customer demand and plan marketing and
inventory strategies accordingly.'''
```

```python
import seaborn as sns
```

```python
#Quantity Distribuition
plt.figure(figsize=(8,5))
sns.histplot(df['Quantity'], bins=50)
plt.title('Quantity Distribution')
plt.show()
'''Insight: Quantity Distribution

The quantity distribution is highly concentrated around zero, with a few extreme positive and negative
values. This indicates the presence of outliers and unusual transactions. Negative quantities may represent
product returns, while very high quantities may indicate bulk purchases.

Recommendation: Investigate extreme values and handle outliers carefully without removing valid return transactions.'''
```

```python
#Top Custommers
top_customers = (
    df.groupby('CustomerID')['TotalSales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_customers)
'''Customer ID 14646 is the highest-value customer, generating total sales of 202,119.21, 
followed by Customer ID 14911 with 108,180.29.
This indicates that a small group of customers contributes significantly to overall sales.'''
```

```python
#Correlation Heatmap
plt.figure(figsize=(8,5))

sns.heatmap(
    df[['Quantity', 'UnitPrice', 'TotalSales']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
'''Insight: Correlation Heatmap

The heatmap shows a strong positive correlation (0.91) between Quantity and TotalSales, indicating
that higher quantities sold generally lead to higher sales. UnitPrice has a moderate positive
correlation (0.40) with TotalSales, while Quantity and UnitPrice have almost no correlation (-0.0012).
Recommendation: Focus on increasing sales volume and maintaining competitive product pricing to
improve overall revenue.'''
```
